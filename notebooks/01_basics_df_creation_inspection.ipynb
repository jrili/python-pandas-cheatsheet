# Python Pandas Cheatsheet

```python
import pandas as pd
```

## Creating Dataframes

### From dictionary of lists (Column-oriented)

```python
# input data: dict of lists; 1 dict member = 1 column
users_data = {
    'user_id': [101, 102, 103, 104],
    'amount': [250.50, 120.00, None, 450.75],
    'status': ['completed', 'pending', 'failed', 'completed']
}

df = pd.DataFrame(users_data)
df
```

### From list of dictionaries (Row-oriented)


```python
# input data: list of dicts; 1 dict = 1 row
users_data = [
    {'user_id':101, 'amount':250.50, 'status':'completed'},
    {'user_id':102, 'amount':120.00, 'status':'pending'},
    {'user_id':103, 'amount':None, 'status':'failed'},
    {'user_id':104, 'amount':450.75, 'status':'completed'}
]

df = pd.DataFrame(users_data)
df
```

### From list of lists (Row-oriented)

```python
# input data: list of lists; 1 list = 1 row
users_data = [
    [101, 250.50, 'completed'],
    [102, 120.00, 'pending'],
    [103, None, 'failed'],
    [104, 450.75, 'completed']
]

# indicate the column names
column_names = ['user_id', 'amount', 'status']

# create the dataframe object
df = pd.DataFrame(users_data, columns=column_names)
df
```
